```python
import pandas as pd
import numpy as np
columns = [
    "age",
    "sex",
    "cp",
    "trestbps",
    "chol",
    "fbs",
    "restecg",
    "thalach",
    "exang",
    "oldpeak",
    "slope",
    "ca",
    "thal",
    "target"
]
df = pd.read_csv(
    "C:/Users/skris/Downloads/AI_Healthcare_Heart_Disease/datasets/processed.cleveland.data",
    names=columns,
    na_values="?"
)
print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())
print("\nLast 5 rows:")
display(df.tail())
```

```python
print(df.dtypes)
```

```python
for column in columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
```

```python
print("\nMissing values BEFORE cleaning:")
print(df.isnull().sum())
```

```python
df["ca"] = df["ca"].fillna(df["ca"].mode()[0])
df["thal"] = df["thal"].fillna(df["thal"].mode()[0])
print("\nMissing values AFTER filling ca and thal:")
print(df.isnull().sum())
```

```python
if df.isnull().sum().sum() > 0:
    print("\nWARNING: Some missing values still exist.")
else:
    print("\nNo missing values remaining.")
```

```python
print("\nDuplicate rows:", df.duplicated().sum())
df = df.drop_duplicates().reset_index(drop=True)
```

```python
integer_columns = [
    "age",
    "sex",
    "cp",
    "fbs",
    "restecg",
    "exang",
    "slope",
    "ca",
    "thal"
]
for column in integer_columns:
    df[column] = df[column].astype(int)
```

```python
df["target"] = df["target"].apply(
    lambda x: 1 if x > 0 else 0
)
df["target"] = df["target"].astype(int)
```

```python
df["disease_status"] = df["target"].map({
    0: "No Heart Disease",
    1: "Heart Disease"
})
```

```python
df = df[
    [
        "age",
        "sex",
        "cp",
        "trestbps",
        "chol",
        "fbs",
        "restecg",
        "thalach",
        "exang",
        "oldpeak",
        "slope",
        "ca",
        "thal",
        "target",
        "disease_status"
    ]
]
```

```python
print("\n" + "=" * 60)
print("FINAL CLEAN DATASET")
print("=" * 60)

print("\nShape:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nTarget distribution:")
print(df["target"].value_counts())

print("\nDisease status:")
print(df["disease_status"].value_counts())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
csv_path = "C:/Users/skris/Downloads/AI_Healthcare_Heart_Disease/datasets/heart_disease_clean.csv"
df.to_csv(
    csv_path,
    index=False
)
print("\nCSV saved successfully:")
print(csv_path)
```
