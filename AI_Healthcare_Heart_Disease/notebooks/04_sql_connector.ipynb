```python
from sqlalchemy import create_engine
username = "root"
password = "shobana07!"
host = "localhost"
database = "heart_disease_db"
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}/{database}"
)
print("Python connected to heart_disease_db!")
```

```python
import pandas as pd
df = pd.read_csv("C:/Users/skris/Downloads/AI_Healthcare_Heart_Disease/datasets/heart_disease_clean.csv")
print("Dataset loaded!")
print("Shape:", df.shape)
display(df.head())
```

```python
df.to_sql(
    name="heart_patients",
    con=engine,
    if_exists="replace",
    index=False
)
print("Data successfully uploaded to MySQL!")
```

```python
query = """
SELECT COUNT(*) AS total_patients
FROM heart_patients;
"""
result = pd.read_sql(query, engine)
display(result)
```

```python
query = """
SELECT 
    target,
    COUNT(*) AS patient_count
FROM heart_patients
GROUP BY target
ORDER BY target;
"""
result = pd.read_sql(query, engine)
display(result)
```
